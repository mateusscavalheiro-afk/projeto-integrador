```python
import pandas as pd
import seaborn as srn
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv( r"C:\Users\mateus_s_cavalheiro\Documents\GitHub\projeto-integrador\dados_limpos.csv")
dataset.head()
```

```python
colunas_interesse = ['Temperatura_C', 'Vibracao_mms', 'Pressao_bar', 'Falha_motor']
matriz_corr = dataset[colunas_interesse].corr()
print(matriz_corr)
```

```python
plt.figure(figsize=(6, 4))
srn.heatmap(matriz_corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Matriz de Correlação')
plt.show()
```
